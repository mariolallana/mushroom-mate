```python
from meteostat import Point, Daily
from datetime import datetime, timedelta

WEATHER_DAYS = 5

end_date = datetime.now().replace(hour=0, minute=0, second=0, microsecond=0)  # Today at midnight
start_date = end_date - timedelta(days=WEATHER_DAYS)

location = Point(40.8319,-3.75816)

print(f"Location: {location}")
print(f"Start date: {start_date}")
print(f"End date: {end_date}")
```

```python
data = Daily(location, start_date, end_date)
df = data.fetch()

print("\nWeather Data:")
print(df.to_string())

# If you want a more concise summary, you can use:
print("\nWeather Data Summary:")
print(df.describe())
```

```python
from meteostat import Stations
import pandas as pd

# Define the bounding box for Comunidad de Madrid
# Approximate coordinates: (40.0, -4.5) to (41.2, -3.0)
bottom_left = (40.0, -4.5)
top_right = (41.2, -3.0)

# Fetch stations within the bounding box
stations = Stations()
stations = stations.bounds(bottom_left, top_right)
stations = stations.fetch()

# Convert to DataFrame and sort by name
df = pd.DataFrame(stations)
df = df.sort_values('name')

# Display the results
print(f"Number of stations found: {len(df)}")
print("\nList of stations in Comunidad de Madrid:")
for _, station in df.iterrows():
    print(f"Name: {station['name']}, ID: {station['id']}, Latitude: {station['latitude']:.4f}, Longitude: {station['longitude']:.4f}")

```

```python
from datetime import datetime, timedelta
from meteostat import Daily

# Step 1: Define the station IDs
station_ids = ['08210', '08215']

# Step 2: Define the date range (last 7 days)
end = datetime.now()
start = end - timedelta(days=7)

# Step 3: Fetch and print the weather data for each station
for station_id in station_ids:
    # Fetch daily weather data for the station
    data = Daily(station_id, start, end)
    data = data.fetch()
    
    # Output the data for the station
    print(f"Weather data for station {station_id}:")
    print(data)
    print("\n")
```
